```python
import numpy as np

print(df.columns.tolist())

print(df.head(2))


df.columns = df.columns.str.strip().str.lower()

credit_col = None
for col in df.columns:
    if 'credit' in col or 'score' in col:
        credit_col = col
        break

df['credit_grp'] = np.where(
    df[credit_col].between(300, 500, inclusive='both'), 'low',
    np.where(
        df[credit_col].between(501, 599, inclusive='both'), 'mid', 
        'high'
    )
)

                                                        
```

```python
print("\n=== RESULT ===")
print(df[[credit_col, 'credit_grp']].head())
print("\n=== COUNTS ===")
print(df['credit_grp'].value_counts())


```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x='credit_grp', hue='loan_status', data=df)
plt.show()
```
